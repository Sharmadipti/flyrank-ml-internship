## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*
The action queue is designed to help content and SEO teams prioritize which pages should be reviewed first.

Priority actions are assigned using observed search-performance signals and the model's risk assessment. The recommendations are intended for human review and should not be interpreted as automatic refresh decisions.

Action categories:

Review: Pages showing signs of decline or elevated went_dark risk.
Improve: Pages with visible search demand but weak performance signals.
Monitor: Pages that remain stable but should continue to be tracked.
Protect: High-performing pages that should avoid unnecessary changes.
Reason codes explain why a page was prioritized. Examples include declining clicks, low CTR, aging content, or elevated model risk.
